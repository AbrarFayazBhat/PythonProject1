```python
import openpyxl
import math
```

TECH PRODUCTIVITY INDEX

```python
def calculate_TPI(filename,sheet_name,column,valid_days):
    wb=openpyxl.load_workbook(filename, read_only=True)
    ws=wb[sheet_name]
    values=[]
    for row in range(6,ws.max_row+1):
        value=ws.cell(row,column).value
        if value is not None:
            values.append(value)
    return math.fsum(values)/valid_days
```

ACADEMIC ACTIVITY INDEX

```python
def calculate_AAI(filename,sheet_name,study_column,class_column,valid_days):
    wb=openpyxl.load_workbook(filename, read_only=True)
    ws=wb[sheet_name]
    study_minutes=[]
    for row in range(6,ws.max_row+1):
        value=ws.cell(row,study_column).value
        if value is not None:
            study_minutes.append(value)
    class_minutes=[]
    for row in range(6,ws.max_row+1):
        value=ws.cell(row,class_column).value
        print(value)
        if value is not None:
            class_minutes.append(value)
    
    return math.fsum(study_minutes+class_minutes)/valid_days
```

```python
tpi=calculate_TPI("12607000.xlsx","Daily Log",5,25)
print(f'Tech Productivity Index is: {tpi}')

aai=calculate_AAI("12607000.xlsx","Daily Log",5,6,25)
print(f'Academic Activity Index is: {aai}')
```
